```python
#Accessing & slice in 2D Array
import numpy as np

a = np.array([[1,2],[3,4]])
print(a)
print(a[1][1])
print(a[0][1])
```

```python
#mathematical operation (Vectorization)

a = np.array([1,2,3])
b = np.array([4,5,6])

print(a+b)
print(a-b)
print(a*b)
print(a**2)
```

```python
#stastical functions

a = np.array([1,2,3,4,5,6])

print(np.max(a))
print(np.min(a))
print(np.sum(a))
print(np.std(a))
print(np.median(a))
```

```python
#Matrix Operations

a = np.array([[1,2],[4,5]])
b = np.array([[4,5,6],[7,8,9]])

print(np.dot(a,b))
print(a.T)
```

```python
#Aggregation

a = np.array([[1,2,3],[4,5,6]])

print(np.sum(a, axis = 1))
print(np.sum(a,axis = 0))
```

```python
#Boolean Filtering

a = np.array([10,20,30,40,50,60,70])

print(a[a>40])
```
