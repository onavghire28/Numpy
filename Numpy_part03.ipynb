```python
#Inserting

import numpy as np

a = np.array([1,2,3,4])
b = np.insert(a,2,5)
print(b)
print(a)
```

```python
#Appending

a = np.array([1,2,3,4,5])

b = np.append(a,[6,7])
print(b)
```

```python
#Concatenate

a = np.array([1,2,3])
b = np.array([4,5,6])

c = np.concatenate((a,b))
print(c)
```

```python
#Deleted

a = np.array([1,2,3,4])

b = np.delete(a,2)
print(b)
```

```python
#Staking
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.vstack((a,b)))
print(np.hstack((a,b)))
```

```python
#splitting

a = np.array([1,2,3,4,5,6])

print(np.split(a,2))
```

```python
#Broadcasting

a = np.array([1,2,3,4,5])
b = a*100
print(a)
print(b)
```
